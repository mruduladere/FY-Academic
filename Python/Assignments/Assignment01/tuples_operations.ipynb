```python
#-------------Operations on Tuple-------------#
```

```python
numbers = (30, 10, 50, 20, 40) 
print("Original Tuple:", numbers)
```

```python
# method 1 : COUNT
count = numbers.count(10)
print("Count of 10:", count)
```

```python
# method 2 : INDEX
index = numbers.index(20)
print("Index of 20:", index)
```

```python
# method 3 : LENGTH
length = len(numbers)
print("Length of tuple:", length)
```

```python
# method 4 : SLICING
print("Tuple after slicing:", numbers[1:4])
```

```python
# method 5 : REPETITION
repeated_tuple = numbers * 2
print("Repeated tuple:", repeated_tuple)
```
