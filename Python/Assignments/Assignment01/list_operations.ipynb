```python
#-------------Operations on list-------------#
```

```python
numbers = [30, 10, 50, 20, 40]   
print("Original List:", numbers)
```

```python
# method 1 : APPEND 
numbers.append(60) 
print("After append:", numbers)
```

```python
# method 2 : INSERT 
numbers.insert(1, 15)
print("After insert:", numbers)
```

```python
# method 3 : REMOVE 
numbers.remove(50)
print("After remove:", numbers)
```

```python
# method 4 : POP 
removed_element = numbers.pop()
print("After pop:", numbers , "Popped element:", removed_element,) 
```

```python
# method 5 : SORT 
numbers.sort()
print("After sorting:", numbers)
```
