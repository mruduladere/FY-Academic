```python
numbers = []

for i in range(20):
    num = int(input("Enter value:"))
    numbers.append(num)

print("List:", numbers)

#a)
print("Similar Elements and their Index :")

for i in range(20):
    if numbers.count(numbers[i])>1:
        print(f"\nChecking element {numbers[i]} at index {i}:")

    for j in range(20):
        if numbers[i]==numbers[j]:
            print(f"Matches at index: {j}")


#b)
even_count = 0
odd_count = 0

for num in numbers:
       if num%2==0:
           even_count+=1
       else:
           odd_count+=1

print("Even values:", even_count)
print("Odd values:", odd_count)

#c)
positive_count = 0
negative_count = 0

for num in numbers:
       if num%2==0:
           positive_count+=1
       else:
           negative_count+=1

print("Positive values:", positive_count)
print("Negative values:", negative_count)

```
