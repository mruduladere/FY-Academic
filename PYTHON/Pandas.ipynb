```python
#Dataframe

import pandas as pd

data = {
    "calories": [420,380,390],
    "duration": [50,40,45]
}

print(pd.DataFrame(data))
```

```python
#Locate Row

import pandas as pd

data = {
       "calories": [420,380,390],
       "duration": [50,40,45] 
}

df = pd.DataFrame(data)
print(df.loc[0])
print(df.loc[[0,1]])
```

```python
#Data Cleaning
import pandas as pd

data = {
    'A': [1,2,3,None,5],
    'B': [None,2,3,4,5],
    'C': [1,2,None,None,5]
}

df = pd.DataFrame(data)
print("Original data:\n",df)

df_cleaned = df.dropna()#deletes rows with missing values
print("Cleaned data:\n", df_cleaned)
```

```python
#Filling Missing Values with 0

import pandas as pd

data = {
    'A': [1,2,3,None,5],
    'B': [None,2,3,4,5],
    'C': [1,2,None,None,5]
}

df = pd.DataFrame(data)
print("Original data:\n",df)

df.fillna(0, inplace = True)
print(df)
```

```python
#Filling Missing Values with other values(here, the mean of the other values in the column)

import pandas as pd

data = {
    'A': [1,2,3,None,5],
    'B': [None,2,3,4,5],
    'C': [1,2,None,None,5]
}

df = pd.DataFrame(data)
print("Original data:\n",df)

df.fillna(df.mean(), inplace = True)
print(df)
```
